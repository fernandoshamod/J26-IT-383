## Architecture, loss and evaluation

YOLOv8n predicts bounding boxes and class scores. Its recorded training uses box localization, classification and distribution-focal losses. Curves are original CSV observations, not recomputed training. Best validation checkpoint selection precedes the reported diagnostic evaluation. AP/mAP integrates precision-recall at declared IoU thresholds; it is not a percentage of correct physical inventory. Missing class support does not become zero-error evidence.
