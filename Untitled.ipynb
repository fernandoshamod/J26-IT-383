```python
import csv
import json
from pathlib import Path
from collections import Counter

# Input: full business JSON file
SOURCE = Path(
    r"D:\Yelp-JSON (1)\Yelp JSON\yelp_dataset"
    r"\yelp_academic_dataset_business.json"
)

# Output: filtered CSV file
OUTPUT = Path(
    r"D:\Vendor-Intelligence-System\event_vendor_candidates.csv"
)

# Proposed category mapping: candidates need manual checking
SERVICE_CATEGORIES = {
    "Catering": {"Caterers"},
    "Photography": {
        "Photographers",
        "Event Photography",
        "Videographers",
    },
    "Decoration": {
        "Florists",
        "Floral Designers",
        "Balloon Services",
    },
    "Entertainment": {
        "DJs",
        "Musicians",
        "Party Characters",
    },
    "Audio Visual": {"Audio/Visual Equipment Rental"},
    "Equipment Rental": {"Party Equipment Rentals"},
    "Event Planning": {
        "Party & Event Planning",
        "Wedding Planning",
    },
}

FIELDS = [
    "business_id",
    "name",
    "service_categories",
    "matched_categories",
    "categories",
    "city",
    "state",
    "latitude",
    "longitude",
    "stars",
    "review_count",
    "is_open",
    "caters",
    "price_range",
    "attributes",
    "hours",
]

# Check input before creating output
if not SOURCE.is_file():
    raise FileNotFoundError(
        f"JSON file not found. Correct the SOURCE path:\n{SOURCE}"
    )

OUTPUT.parent.mkdir(parents=True, exist_ok=True)

scanned = 0
saved = 0
seen_ids = set()
service_counts = Counter()

with SOURCE.open("r", encoding="utf-8-sig") as infile:
    with OUTPUT.open(
        "w", newline="", encoding="utf-8-sig"
    ) as outfile:
        writer = csv.DictWriter(outfile, fieldnames=FIELDS)
        writer.writeheader()

        for line_number, line in enumerate(infile, start=1):
            if not line.strip():
                continue

            try:
                business = json.loads(line)
            except json.JSONDecodeError as error:
                raise ValueError(
                    f"Invalid JSON at line {line_number}"
                ) from error

            scanned += 1

            # Dataset open status, NOT booking availability
            if business.get("is_open") != 1:
                continue

            categories = {
                item.strip()
                for item in (
                    business.get("categories") or ""
                ).split(",")
                if item.strip()
            }

            attributes = business.get("attributes") or {}
            services = set()
            matched = set()

            for service, allowed in SERVICE_CATEGORIES.items():
                matches = categories & allowed

                if matches:
                    services.add(service)
                    matched.update(matches)

            # Food businesses explicitly marked as offering catering
            caters = attributes.get("Caters")

            if (
                str(caters).strip().lower() == "true"
                and categories & {"Restaurants", "Food"}
            ):
                services.add("Catering")
                matched.add("Caters=True")

            if not services:
                continue

            business_id = business.get("business_id")

            if not business_id or business_id in seen_ids:
                continue

            seen_ids.add(business_id)

            row = {
                field: business.get(field)
                for field in FIELDS
            }

            row["service_categories"] = "; ".join(sorted(services))
            row["matched_categories"] = "; ".join(sorted(matched))
            row["caters"] = caters
            row["price_range"] = attributes.get(
                "RestaurantsPriceRange2"
            )

            for field in ("attributes", "hours"):
                value = business.get(field)
                row[field] = (
                    json.dumps(value, ensure_ascii=False)
                    if value is not None else ""
                )

            writer.writerow(row)
            service_counts.update(services)
            saved += 1

# Check CSV row count and collect a small preview
preview = []
output_rows = 0

with OUTPUT.open(
    "r", newline="", encoding="utf-8-sig"
) as checkfile:
    reader = csv.DictReader(checkfile)

    for row in reader:
        output_rows += 1

        if len(preview) < 5:
            preview.append(row)

assert output_rows == saved, "CSV row count mismatch"

print("Completed!")
print(f"Business records scanned: {scanned:,}")
print(f"Unique vendor candidates saved: {saved:,}")
print(f"CSV saved to: {OUTPUT}")

print("\nCandidates per service:")
for service in SERVICE_CATEGORIES:
    print(f"{service}: {service_counts,}")

print("\nOne vendor may belong to multiple services.")

print("\nFirst five candidates:")
for row in preview:
    print(
        row["name"],
        "|",
        row["service_categories"],
        "|",
        row["city"]
    )

if saved == 0:
    print("\nNo matching candidates found. Check the source data.")
```

```python
import csv
import json
from pathlib import Path

# Input files
vendor_csv = Path(
    r"D:\Vendor-Intelligence-System\event_vendor_candidates.csv"
)

review_json = Path(
    r"D:\Yelp-JSON (1)\Yelp JSON\yelp_dataset\yelp_academic_dataset_review.json"
)

# Output file
output_csv = Path(
    r"D:\Vendor-Intelligence-System\event_vendor_reviews.csv"
)

temporary_csv = output_csv.with_suffix(".csv.tmp")

# Check input files
for file_path in (vendor_csv, review_json):
    if not file_path.is_file():
        raise FileNotFoundError(
            "File not found: " + str(file_path)
        )

# Read selected vendor IDs
vendor_ids = set()

with vendor_csv.open(
    "r", encoding="utf-8-sig", newline=""
) as file:
    reader = csv.DictReader(file)

    if "business_id" not in (reader.fieldnames or []):
        raise ValueError("Vendor CSV has no business_id column.")

    for row in reader:
        business_id = (row.get("business_id") or "").strip()

        if business_id:
            vendor_ids.add(business_id)

if not vendor_ids:
    raise ValueError("No business IDs found in vendor CSV.")

columns = [
    "review_id",
    "business_id",
    "stars",
    "text",
    "date",
    "useful",
    "funny",
    "cool",
]

output_csv.parent.mkdir(parents=True, exist_ok=True)

scanned = 0
saved = 0
matched_vendor_ids = set()

print("Selected vendors:", len(vendor_ids))

# Read full review JSON one line at a time
with review_json.open("r", encoding="utf-8-sig") as source:
    with temporary_csv.open(
        "w", encoding="utf-8-sig", newline=""
    ) as destination:
        writer = csv.DictWriter(destination, fieldnames=columns)
        writer.writeheader()

        for line_number, line in enumerate(source, start=1):
            if not line.strip():
                continue

            try:
                review = json.loads(line)
            except json.JSONDecodeError as error:
                message = "Invalid JSON at line " + str(line_number)
                raise ValueError(message) from error

            scanned += 1
            business_id = review.get("business_id")

            if business_id in vendor_ids:
                output_row = {
                    column: review.get(column)
                    for column in columns
                }

                writer.writerow(output_row)
                matched_vendor_ids.add(business_id)
                saved += 1

            if scanned % 1000000 == 0:
                print(
                    "Scanned:", scanned,
                    "| Matching reviews:", saved
                )

# Check the completed CSV
checked_rows = 0
preview = []

with temporary_csv.open(
    "r", encoding="utf-8-sig", newline=""
) as file:
    reader = csv.DictReader(file)

    for row in reader:
        if row["business_id"] not in vendor_ids:
            raise ValueError("Unexpected business ID in output.")

        checked_rows += 1

        if len(preview) < 5:
            preview.append({
                "business_id": row["business_id"],
                "stars": row["stars"],
                "date": row["date"],
            })

if checked_rows != saved:
    raise ValueError("Output row count mismatch.")

# Replace output only after successful conversion
temporary_csv.replace(output_csv)

print("\nCOMPLETED")
print("Review records scanned:", scanned)
print("Matching reviews saved:", saved)
print("Vendors with reviews:", len(matched_vendor_ids))
print(
    "Vendors without matching reviews:",
    len(vendor_ids - matched_vendor_ids)
)
print("CSV saved to:", str(output_csv))

print("\nFirst five matching records:")
for row in preview:
    print(row)

if saved == 0:
    print("No matching reviews found. Check the input files.")
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

PROJECT = Path(r"D:\Vendor-Intelligence-System")

VENDOR_FILE = PROJECT / "event_vendor_candidates.csv"
REVIEW_FILE = PROJECT / "event_vendor_reviews.csv"
OUTPUT_FILE = PROJECT / "vendor_features.csv"

for file in (VENDOR_FILE, REVIEW_FILE):
    if not file.is_file():
        raise FileNotFoundError("File not found: " + str(file))

vendors = pd.read_csv(
    VENDOR_FILE,
    encoding="utf-8-sig",
    dtype={"business_id": "string"}
)

if vendors["business_id"].isna().any():
    raise ValueError("Missing vendor business IDs.")

if vendors["business_id"].duplicated().any():
    raise ValueError("Duplicate vendor business IDs.")

print("Vendor records:", len(vendors))
display(vendors.head())
```

```python
parts = []
processed = 0
vendor_ids = set(vendors["business_id"])

chunks = pd.read_csv(
    REVIEW_FILE,
    usecols=["business_id", "stars", "date"],
    dtype={"business_id": "string"},
    encoding="utf-8-sig",
    chunksize=50000
)

for chunk in chunks:
    if not chunk["business_id"].isin(vendor_ids).all():
        raise ValueError("Reviews contain an unexpected business ID.")

    chunk["stars"] = pd.to_numeric(
        chunk["stars"], errors="coerce"
    )

    if (
        chunk["stars"].isna().any()
        or not chunk["stars"].between(1, 5).all()
    ):
        raise ValueError("Missing or invalid review ratings.")

    chunk["date"] = pd.to_datetime(
        chunk["date"], errors="coerce"
    )

    if chunk["date"].isna().any():
        raise ValueError("Missing or invalid review dates.")

    chunk["rating_squared"] = chunk["stars"] ** 2

    summary = chunk.groupby("business_id").agg(
        collected_review_count=("stars", "size"),
        rating_sum=("stars", "sum"),
        rating_squared_sum=("rating_squared", "sum"),
        latest_review_date=("date", "max")
    )

    parts.append(summary)
    processed += len(chunk)

    if processed % 500000 == 0:
        print("Reviews processed:", processed)

if not parts:
    raise ValueError("Review file contains no records.")

# Combine summaries from all chunks
review_summary = pd.concat(parts).groupby(level=0).agg({
    "collected_review_count": "sum",
    "rating_sum": "sum",
    "rating_squared_sum": "sum",
    "latest_review_date": "max"
})

count = review_summary["collected_review_count"]

review_summary["mean_review_rating"] = (
    review_summary["rating_sum"] / count
)

# Sample standard deviation.
# A vendor with only one review has undefined standard deviation.
variance = (
    review_summary["rating_squared_sum"]
    - (review_summary["rating_sum"] ** 2 / count)
) / (count - 1).replace(0, np.nan)

review_summary["rating_std"] = np.sqrt(
    variance.clip(lower=0)
)

review_summary = review_summary[
    [
        "collected_review_count",
        "mean_review_rating",
        "rating_std",
        "latest_review_date"
    ]
].reset_index()

assert (
    review_summary["collected_review_count"].sum()
    == processed
), "Review count mismatch."

print("Total reviews processed:", processed)
print("Vendors with review summaries:", len(review_summary))
display(review_summary.head())
```

```python
combined = vendors.merge(
    review_summary,
    on="business_id",
    how="left",
    validate="one_to_one"
)

combined["has_review_data"] = (
    combined["collected_review_count"].notna()
)

combined["collected_review_count"] = (
    combined["collected_review_count"]
    .fillna(0)
    .astype("int64")
)

assert len(combined) == len(vendors)
assert not combined["business_id"].duplicated().any()

combined.to_csv(
    OUTPUT_FILE,
    index=False,
    encoding="utf-8-sig"
)

# Read saved CSV to check record count
saved = pd.read_csv(
    OUTPUT_FILE,
    encoding="utf-8-sig",
    dtype={"business_id": "string"}
)

assert len(saved) == len(vendors)
assert not saved["business_id"].duplicated().any()

print("COMPLETED")
print("Vendor rows:", len(saved))
print("Columns:", len(saved.columns))
print(
    "Vendors without review data:",
    int((~combined["has_review_data"]).sum())
)
print("Saved to:", str(OUTPUT_FILE))

display(saved.head())

print("\nMissing values:")
print(saved.isna().sum())
```

```python
import pandas as pd
from pathlib import Path

PROJECT = Path(r"D:\Vendor-Intelligence-System")

SOURCE = PROJECT / "vendor_features.csv"
OUTPUT = PROJECT / "vendor_service_details_prefilled.csv"

if not SOURCE.is_file():
    raise FileNotFoundError(
        "File not found: " + str(SOURCE)
    )

# Read vendor features
vendors = pd.read_csv(
    SOURCE,
    encoding="utf-8-sig",
    dtype={"business_id": "string"}
)

required = [
    "business_id",
    "name",
    "service_categories",
    "city",
    "state",
    "price_range",
]

missing_columns = [
    column for column in required
    if column not in vendors.columns
]

if missing_columns:
    raise ValueError(
        "Missing columns: " + ", ".join(missing_columns)
    )

if vendors["business_id"].isna().any():
    raise ValueError("Missing business IDs.")

if vendors["business_id"].duplicated().any():
    raise ValueError("Duplicate business IDs.")

if vendors["service_categories"].isna().any():
    raise ValueError("Missing service categories.")

# One row per vendor-service combination
details = vendors[required].copy()

details["service_category"] = (
    details["service_categories"]
    .str.split(";")
)

details = details.explode(
    "service_category",
    ignore_index=True
)

details["service_category"] = (
    details["service_category"].str.strip()
)

details = details[
    details["service_category"].ne("")
].copy()

details = details.rename(columns={
    "name": "vendor_name",
    "city": "business_city",
    "state": "business_state",
    "price_range": "yelp_price_range_indicator",
})

# Unknown fields stay blank
unknown_columns = [
    "package_price",
    "currency",
    "pricing_basis",
    "guest_capacity",
    "service_area",
    "backup_capability",
]

for column in unknown_columns:
    details[column] = pd.NA

details["data_source"] = (
    "Yelp business data via vendor_features.csv"
)

details["verification_status"] = (
    "Candidate service mapping; event capability unverified"
)

columns = [
    "business_id",
    "service_category",
    "package_price",
    "currency",
    "pricing_basis",
    "guest_capacity",
    "service_area",
    "backup_capability",
    "data_source",
    "vendor_name",
    "business_city",
    "business_state",
    "yelp_price_range_indicator",
    "verification_status",
]

details = details[columns]

if details.duplicated(
    ["business_id", "service_category"]
).any():
    raise ValueError("Duplicate vendor-service pairs.")

# Save a separate prefilled file
details.to_csv(
    OUTPUT,
    index=False,
    encoding="utf-8-sig"
)

# Check saved file
saved = pd.read_csv(
    OUTPUT,
    encoding="utf-8-sig",
    dtype={"business_id": "string"}
)

assert len(saved) == len(details)

assert set(saved["business_id"]) == set(
    vendors["business_id"]
)

assert not saved.duplicated(
    ["business_id", "service_category"]
).any()

print("COMPLETED")
print("Unique vendors:", saved["business_id"].nunique())
print("Vendor-service rows:", len(saved))
print("Saved to:", str(OUTPUT))

print("\nRows per service:")
print(saved["service_category"].value_counts())

display(saved.head(10))
```

```python
import pandas as pd

df = pd.read_csv(
    r"D:\Vendor-Intelligence-System\vendor_features.csv"
)

def risk_label(row):

    if (
        row["mean_review_rating"] >= 4
        and row["rating_std"] < 1
        and row["collected_review_count"] >= 20
    ):
        return "Low"

    elif row["mean_review_rating"] >= 3:
        return "Medium"

    else:
        return "High"

df["risk_level"] = df.apply(
    risk_label,
    axis=1
)

df.to_csv(
    r"D:\Vendor-Intelligence-System\vendor_risk_dataset.csv",
    index=False
)

print(df["risk_level"].value_counts())
```

```python
def get_vendor(vendor_name):

    row = vendors[
        vendors["name"] == vendor_name
    ]

    if len(row) == 0:
        return None

    return row.iloc[0]


def compatibility_score(
    vendor_a_name,
    vendor_b_name
):

    vendor_a = get_vendor(vendor_a_name)
    vendor_b = get_vendor(vendor_b_name)

    if vendor_a is None:
        return "Vendor A not found"

    if vendor_b is None:
        return "Vendor B not found"

    service_a = str(
        vendor_a["service_categories"]
    ).split(";")[0].strip()

    service_b = str(
        vendor_b["service_categories"]
    ).split(";")[0].strip()

    match = matrix[
        (
            (matrix["service_1"] == service_a)
            &
            (matrix["service_2"] == service_b)
        )
        |
        (
            (matrix["service_1"] == service_b)
            &
            (matrix["service_2"] == service_a)
        )
    ]

    if len(match) > 0:
        category_score = (
            match.iloc[0]["base_score"]
        )
    else:
        category_score = 50

    risk_bonus = 0

    if (
        vendor_a["risk_level"] == "Low"
        and
        vendor_b["risk_level"] == "Low"
    ):
        risk_bonus = 20

    elif (
        "Low"
        in [
            vendor_a["risk_level"],
            vendor_b["risk_level"]
        ]
    ):
        risk_bonus = 10

    score_gap = abs(
        vendor_a["vendor_score"]
        -
        vendor_b["vendor_score"]
    )

    quality_bonus = 0

    if score_gap <= 20:
        quality_bonus = 10

    elif score_gap <= 40:
        quality_bonus = 5

    final_score = min(
        category_score +
        risk_bonus +
        quality_bonus,
        100
    )

    if final_score >= 90:
        level = "Highly Compatible"

    elif final_score >= 70:
        level = "Compatible"

    elif final_score >= 50:
        level = "Moderate"

    else:
        level = "Conflict"

    return {
        "Vendor A": vendor_a_name,
        "Vendor B": vendor_b_name,
        "Service A": service_a,
        "Service B": service_b,
        "Compatibility Score": final_score,
        "Compatibility Level": level
    }
```

```python
import pandas as pd
from pathlib import Path

DATA_FILE = Path(
    r"D:\Vendor-Intelligence-System\vendor_recommendation_dataset.csv"
)

# Example location from your available locations list
EVENT_CITY = "Alton"
EVENT_STATE = "IL"

if not DATA_FILE.is_file():
    raise FileNotFoundError("File not found: " + str(DATA_FILE))

package_vendors = pd.read_csv(
    DATA_FILE,
    encoding="utf-8-sig",
    dtype={"business_id": "string"}
)

required_columns = [
    "business_id",
    "name",
    "city",
    "state",
    "service_categories",
    "vendor_score",
    "risk_level",
]

for column in required_columns:
    if column not in package_vendors.columns:
        raise ValueError("Missing column: " + column)

package_vendors["vendor_score"] = pd.to_numeric(
    package_vendors["vendor_score"],
    errors="raise"
)

if package_vendors["vendor_score"].isna().any():
    raise ValueError("Missing vendor scores.")


def generate_local_package(city, state):
    services = ["Catering", "Photography", "Decoration"]

    city_match = (
        package_vendors["city"]
        .fillna("")
        .str.strip()
        .str.casefold()
        .eq(city.strip().casefold())
    )

    state_match = (
        package_vendors["state"]
        .fillna("")
        .str.strip()
        .str.casefold()
        .eq(state.strip().casefold())
    )

    local_vendors = package_vendors.loc[
        city_match & state_match
    ].copy()

    selected = []
    missing_services = []

    for service in services:
        service_match = (
            local_vendors["service_categories"]
            .fillna("")
            .apply(
                lambda value: service in {
                    item.strip()
                    for item in value.split(";")
                }
            )
        )

        candidates = local_vendors.loc[service_match].sort_values(
            by=["vendor_score", "business_id"],
            ascending=[False, True]
        )

        if candidates.empty:
            missing_services.append(service)
            continue

        vendor = candidates.iloc[0]

        selected.append({
            "Service": service,
            "Business ID": str(vendor["business_id"]),
            "Vendor": vendor["name"],
            "Vendor Score": float(vendor["vendor_score"]),
            "Rule-Based Risk Label": vendor["risk_level"],
        })

    if missing_services:
        return {
            "Status": "Incomplete candidate package",
            "City": city,
            "State": state,
            "Missing Services": missing_services,
        }

    scores = [item["Vendor Score"] for item in selected]
    average_score = round(sum(scores) / len(scores), 2)

    return {
        "Status": "Candidate package only",
        "City": city,
        "State": state,
        "Selected Vendors": selected,
        "Average Vendor Score": average_score,
        "Service Area Check": "Not evaluated",
        "Budget Check": "Not evaluated",
        "Availability Check": "Not evaluated",
        "Venue Check": "Not evaluated",
        "Operational Compatibility": "Not evaluated",
    }


# Call the function after defining it
package = generate_local_package(EVENT_CITY, EVENT_STATE)

print("Status:", package["Status"])
print("Location:", package["City"], "/", package["State"])

if "Selected Vendors" in package:
    display(pd.DataFrame(package["Selected Vendors"]))
    print("Average Vendor Score:", package["Average Vendor Score"])

    for check in [
        "Service Area Check",
        "Budget Check",
        "Availability Check",
        "Venue Check",
        "Operational Compatibility",
    ]:
        print(check + ":", package[check])
else:
    print("Missing Services:", package["Missing Services"])
```

```python
def evaluate_vendor_candidates(requirements, data):
    city = str(requirements.get("city") or "").strip()
    state = str(requirements.get("state") or "").strip()
    services = requirements.get("services") or {}

    if not city or not state:
        raise ValueError("Enter event city and state.")

    if not isinstance(services, dict) or not services:
        raise ValueError("Enter at least one required service.")

    default_weights = {
        "rating": 0.60,
        "review_volume": 0.20,
        "review_consistency": 0.20
    }

    supplied = requirements.get("priorities") or {}
    values = [supplied.get(key) for key in default_weights]

    if all(value is None for value in values):
        weights = default_weights.copy()
        weight_source = "Proposed default weights"
    elif any(value is None for value in values):
        raise ValueError("Enter all three priority weights.")
    else:
        weights = {
            key: float(supplied[key])
            for key in default_weights
        }

        if any(
            not np.isfinite(value) or value < 0
            for value in weights.values()
        ):
            raise ValueError("Weights must be finite and non-negative.")

        total = sum(weights.values())

        if total <= 0:
            raise ValueError("Priority weights cannot all be zero.")

        weights = {
            key: value / total
            for key, value in weights.items()
        }
        weight_source = "User priority weights"

    local = data[
        data["city"].fillna("").str.strip().str.casefold().eq(
            city.casefold()
        )
        & data["state"].fillna("").str.strip().str.casefold().eq(
            state.casefold()
        )
    ].copy()

    if requirements.get("exclude_high_review_labels", True):
        allowed_labels = ["Low", "Medium"]
    else:
        allowed_labels = ["Low", "Medium", "High"]

    local = local[
        local["risk_level"].isin(allowed_labels)
    ].copy()

    numeric_columns = [
        "mean_review_rating",
        "collected_review_count",
        "rating_std"
    ]

    for column in numeric_columns:
        local[column] = pd.to_numeric(
            local[column], errors="coerce"
        )

    valid = (
        local[numeric_columns].notna().all(axis=1)
        & local["mean_review_rating"].between(1, 5)
        & local["collected_review_count"].ge(1)
        & local["rating_std"].ge(0)
        & np.isfinite(local[numeric_columns]).all(axis=1)
    )

    invalid_feature_count = int((~valid).sum())
    local = local.loc[valid].copy()

    results = []
    missing_services = []

    for service in services:
        service = str(service).strip()

        if not service:
            raise ValueError("Required service name cannot be blank.")

        mask = local["service_categories"].fillna("").apply(
            lambda value: service in {
                item.strip() for item in value.split(";")
            }
        )

        candidates = local.loc[mask].copy()

        if candidates.empty:
            missing_services.append(service)
            continue

        candidates["rating_component"] = (
            candidates["mean_review_rating"] / 5
        )

        # Normalise review volume within this local service pool.
        maximum = candidates["collected_review_count"].max()

        candidates["volume_component"] = (
            np.log1p(candidates["collected_review_count"])
            / np.log1p(maximum)
        )

        candidates["consistency_component"] = (
            1 / (1 + candidates["rating_std"])
        )

        candidates["event_ranking_score"] = 100 * (
            weights["rating"] * candidates["rating_component"]
            + weights["review_volume"] * candidates["volume_component"]
            + weights["review_consistency"]
            * candidates["consistency_component"]
        )

        candidates["assigned_service"] = service
        candidates["eligibility_status"] = "Candidate only"

        candidates["selection_reason"] = (
            "Service category and business location match; "
            "review label meets the selected preference."
        )

        candidates["checks_not_evaluated"] = (
            "Event-type suitability; budget; guest capacity; "
            "date/time availability; service-specific needs; "
            "service area; venue; operational compatibility"
        )

        results.append(candidates)

    columns = [
        "business_id", "name", "assigned_service",
        "city", "state", "risk_level",
        "mean_review_rating", "collected_review_count",
        "rating_std", "event_ranking_score",
        "eligibility_status", "selection_reason",
        "checks_not_evaluated"
    ]

    if results:
        evaluated = pd.concat(results, ignore_index=True)

        evaluated = evaluated.sort_values(
            ["assigned_service", "event_ranking_score", "business_id"],
            ascending=[True, False, True]
        )

        evaluated = evaluated[columns].reset_index(drop=True)
    else:
        evaluated = pd.DataFrame(columns=columns)

    return {
        "candidates": evaluated,
        "missing_services": missing_services,
        "weights": weights,
        "weight_source": weight_source,
        "invalid_feature_rows_excluded": invalid_feature_count
    }
```

```python
from itertools import product
import math
import pandas as pd


def generate_dynamic_packages(
    requirements,
    vendor_data,
    top_per_service=3,
    max_combinations=10000
):
    if not isinstance(top_per_service, int) or top_per_service < 1:
        raise ValueError("top_per_service must be a positive integer.")

    number_of_packages = requirements.get("number_of_packages", 3)

    if (
        not isinstance(number_of_packages, int)
        or isinstance(number_of_packages, bool)
        or number_of_packages < 1
    ):
        raise ValueError("number_of_packages must be a positive integer.")

    # Evaluate individual vendors using the previous function
    evaluation = evaluate_vendor_candidates(
        requirements,
        vendor_data
    )

    candidates = evaluation["candidates"]
    services = [
        str(service).strip()
        for service in requirements["services"]
    ]

    if evaluation["missing_services"]:
        return {
            "status": "Incomplete: missing eligible candidates",
            "missing_services": evaluation["missing_services"],
            "packages": [],
            "weights": evaluation["weights"]
        }

    # Create one candidate pool per requested service
    pools = []

    for service in services:
        pool = candidates.loc[
            candidates["assigned_service"].eq(service)
        ].sort_values(
            ["event_ranking_score", "business_id"],
            ascending=[False, True]
        )

        pool = pool.drop_duplicates("business_id")
        pool = pool.head(top_per_service)

        if pool.empty:
            return {
                "status": "Incomplete: missing eligible candidates",
                "missing_services": [service],
                "packages": [],
                "weights": evaluation["weights"]
            }

        pools.append(pool.to_dict("records"))

    combination_count = math.prod(len(pool) for pool in pools)

    if combination_count > max_combinations:
        raise ValueError(
            "Too many combinations. Reduce top_per_service "
            "or use a bounded search method."
        )

    packages = []
    seen = set()
    excluded_shared_vendor_combinations = 0

    for combination in product(*pools):
        business_ids = [
            item["business_id"]
            for item in combination
        ]

        # Conservative prototype policy:
        # one distinct vendor per service until multi-service
        # delivery capability has been verified.
        if len(set(business_ids)) != len(business_ids):
            excluded_shared_vendor_combinations += 1
            continue

        signature = tuple(
            (item["assigned_service"], item["business_id"])
            for item in combination
        )

        if signature in seen:
            continue

        seen.add(signature)

        selected = [
            {
                "service": item["assigned_service"],
                "business_id": item["business_id"],
                "vendor_name": item["name"],
                "review_based_label": item["risk_level"],
                "event_ranking_score": float(
                    item["event_ranking_score"]
                )
            }
            for item in combination
        ]

        # Equal service weights for this prototype
        package_score = sum(
            item["event_ranking_score"]
            for item in selected
        ) / len(selected)

        packages.append({
            "selected_vendors": selected,
            "package_score": package_score,
            "status": "Candidate package only",
            "checks": {
                "required_service_categories": "Matched",
                "business_city_state": "Matched",
                "review_label_preference": "Matched",
                "event_type_suitability": "Not evaluated",
                "budget": "Not evaluated",
                "guest_capacity": "Not evaluated",
                "date_time_availability": "Not evaluated",
                "service_specific_requirements": "Not evaluated",
                "service_area": "Not evaluated",
                "venue": "Not evaluated",
                "operational_compatibility": "Not evaluated"
            },
            "explanation": (
                "Selected from matching service categories and "
                "business locations, using the chosen review-label "
                "preference and ranking priorities. "
                "Other feasibility requirements remain unverified."
            ),
            "_signature": signature
        })

    packages.sort(
        key=lambda item: (
            -item["package_score"],
            item["_signature"]
        )
    )

    best_packages = packages[:number_of_packages]

    for rank, package in enumerate(best_packages, start=1):
        package["package_rank"] = rank
        package["package_score"] = round(
            float(package["package_score"]), 2
        )
        del package["_signature"]

    return {
        "status": (
            "Candidate packages generated"
            if best_packages
            else "No package under distinct-vendor policy"
        ),
        "missing_services": [],
        "packages": best_packages,
        "combinations_considered": combination_count,
        "shared_vendor_combinations_excluded":
            excluded_shared_vendor_combinations,
        "weights": evaluation["weights"],
        "weight_source": evaluation["weight_source"]
    }
```

```python
from itertools import product
import math
import pandas as pd


def generate_dynamic_packages(
    requirements,
    vendor_data,
    top_per_service=3,
    max_combinations=10000
):
    if not isinstance(top_per_service, int) or top_per_service < 1:
        raise ValueError("top_per_service must be a positive integer.")

    number_of_packages = requirements.get("number_of_packages", 3)

    if (
        not isinstance(number_of_packages, int)
        or isinstance(number_of_packages, bool)
        or number_of_packages < 1
    ):
        raise ValueError("number_of_packages must be a positive integer.")

    # Evaluate individual vendors using the previous function
    evaluation = evaluate_vendor_candidates(
        requirements,
        vendor_data
    )

    candidates = evaluation["candidates"]
    services = [
        str(service).strip()
        for service in requirements["services"]
    ]

    if evaluation["missing_services"]:
        return {
            "status": "Incomplete: missing eligible candidates",
            "missing_services": evaluation["missing_services"],
            "packages": [],
            "weights": evaluation["weights"]
        }

    # Create one candidate pool per requested service
    pools = []

    for service in services:
        pool = candidates.loc[
            candidates["assigned_service"].eq(service)
        ].sort_values(
            ["event_ranking_score", "business_id"],
            ascending=[False, True]
        )

        pool = pool.drop_duplicates("business_id")
        pool = pool.head(top_per_service)

        if pool.empty:
            return {
                "status": "Incomplete: missing eligible candidates",
                "missing_services": [service],
                "packages": [],
                "weights": evaluation["weights"]
            }

        pools.append(pool.to_dict("records"))

    combination_count = math.prod(len(pool) for pool in pools)

    if combination_count > max_combinations:
        raise ValueError(
            "Too many combinations. Reduce top_per_service "
            "or use a bounded search method."
        )

    packages = []
    seen = set()
    excluded_shared_vendor_combinations = 0

    for combination in product(*pools):
        business_ids = [
            item["business_id"]
            for item in combination
        ]

        # Conservative prototype policy:
        # one distinct vendor per service until multi-service
        # delivery capability has been verified.
        if len(set(business_ids)) != len(business_ids):
            excluded_shared_vendor_combinations += 1
            continue

        signature = tuple(
            (item["assigned_service"], item["business_id"])
            for item in combination
        )

        if signature in seen:
            continue

        seen.add(signature)

        selected = [
            {
                "service": item["assigned_service"],
                "business_id": item["business_id"],
                "vendor_name": item["name"],
                "review_based_label": item["risk_level"],
                "event_ranking_score": float(
                    item["event_ranking_score"]
                )
            }
            for item in combination
        ]

        # Equal service weights for this prototype
        package_score = sum(
            item["event_ranking_score"]
            for item in selected
        ) / len(selected)

        packages.append({
            "selected_vendors": selected,
            "package_score": package_score,
            "status": "Candidate package only",
            "checks": {
                "required_service_categories": "Matched",
                "business_city_state": "Matched",
                "review_label_preference": "Matched",
                "event_type_suitability": "Not evaluated",
                "budget": "Not evaluated",
                "guest_capacity": "Not evaluated",
                "date_time_availability": "Not evaluated",
                "service_specific_requirements": "Not evaluated",
                "service_area": "Not evaluated",
                "venue": "Not evaluated",
                "operational_compatibility": "Not evaluated"
            },
            "explanation": (
                "Selected from matching service categories and "
                "business locations, using the chosen review-label "
                "preference and ranking priorities. "
                "Other feasibility requirements remain unverified."
            ),
            "_signature": signature
        })

    packages.sort(
        key=lambda item: (
            -item["package_score"],
            item["_signature"]
        )
    )

    best_packages = packages[:number_of_packages]

    for rank, package in enumerate(best_packages, start=1):
        package["package_rank"] = rank
        package["package_score"] = round(
            float(package["package_score"]), 2
        )
        del package["_signature"]

    return {
        "status": (
            "Candidate packages generated"
            if best_packages
            else "No package under distinct-vendor policy"
        ),
        "missing_services": [],
        "packages": best_packages,
        "combinations_considered": combination_count,
        "shared_vendor_combinations_excluded":
            excluded_shared_vendor_combinations,
        "weights": evaluation["weights"],
        "weight_source": evaluation["weight_source"]
    }
```

```python
import pandas as pd
from pathlib import Path

PROJECT = Path(r"D:\Vendor-Intelligence-System")

options = pd.read_csv(
    PROJECT / "gfh_published_service_options.csv",
    encoding="utf-8-sig"
)

conditions = pd.read_csv(
    PROJECT / "gfh_option_conditions.csv",
    encoding="utf-8-sig"
)


def calculate_provisional_cost(
    option_id,
    guest_count,
    vendor_budget,
    event_date,
    hall_name
):
    if (
        not isinstance(guest_count, int)
        or isinstance(guest_count, bool)
        or guest_count <= 0
    ):
        raise ValueError("Guest count must be a positive integer.")

    budget = float(vendor_budget)

    if not pd.notna(budget) or budget <= 0 or budget == float("inf"):
        raise ValueError("Budget must be a finite positive number.")

    date = pd.to_datetime(
        event_date,
        format="%Y-%m-%d",
        errors="raise"
    )

    matched = options[
        options["service_option_id"].eq(option_id)
    ]

    if len(matched) != 1:
        raise ValueError("Option ID must match exactly one record.")

    option = matched.iloc[0]

    if option["currency"] != "LKR":
        raise ValueError("This calculator expects an LKR budget.")

    if option["pricing_basis"] != "per_person":
        raise ValueError("Unsupported pricing basis.")

    price = float(option["unit_price"])

    if not pd.notna(price) or price <= 0 or price == float("inf"):
        raise ValueError("Invalid unit price.")

    base_cost = round(price * guest_count, 2)
    budget_balance = round(budget - base_cost, 2)

    # Expiry alone does not establish the applicable start date.
    expiry = pd.to_datetime(option["valid_until"], errors="raise")

    if date > expiry:
        price_period_check = "Failed: event is after stated expiry"
    elif pd.isna(option["valid_from"]):
        price_period_check = (
            "Unknown: valid_from is missing; rate confirmation required"
        )
    else:
        start = pd.to_datetime(option["valid_from"], errors="raise")

        price_period_check = (
            "Within recorded price period; confirmation required"
            if date >= start
            else "Failed: event is before stated start"
        )

    hall_condition = conditions[
        conditions["service_option_id"].eq(option_id)
        & conditions["hall_name"].eq(hall_name)
    ]

    if len(hall_condition) == 1:
        minimum = int(hall_condition.iloc[0]["minimum_guests"])

        inclusion_check = (
            "Meets recorded inclusion guest-count condition"
            if guest_count >= minimum
            else "Does not meet recorded inclusion guest-count condition"
        )
    else:
        minimum = None
        inclusion_check = "Unknown: no unique hall condition found"

    return {
        "service_option_id": option_id,
        "option_name": option["service_option_name"],
        "guest_count": guest_count,
        "currency": "LKR",
        "published_unit_price": price,
        "provisional_base_cost": base_cost,
        "vendor_budget": budget,
        "budget_balance_before_extras": budget_balance,
        "base_cost_budget_check": (
            "Within budget before extras"
            if base_cost <= budget
            else "Failed: base cost exceeds budget"
        ),
        "price_period_check": price_period_check,
        "inclusion_minimum_guests": minimum,
        "inclusion_condition_check": inclusion_check,
        "cost_completeness": "Base price only; selected extras not evaluated",
        "availability_check": "Unknown",
        "venue_check": "Not evaluated",
        "overall_status": "Provisional calculation, not booking approval"
    }
```

```python
def calculate_selected_addons(base_result, feature_data, selections):
    """
    selections:
    [{"feature_record_id": "...", "quantity": 2}]
    
    Calculates only recorded per_bottle add-ons.
    All costs remain provisional source-text calculations.
    """
    selected_ids = [
        item["feature_record_id"] for item in selections
    ]

    if len(selected_ids) != len(set(selected_ids)):
        raise ValueError("Duplicate add-on selections.")

    rows = []
    known_addon_total = 0.0
    unresolved = []

    for selection in selections:
        feature_id = selection["feature_record_id"]
        quantity = selection["quantity"]

        if (
            isinstance(quantity, bool)
            or not isinstance(quantity, int)
            or quantity <= 0
        ):
            raise ValueError("Quantity must be a positive integer.")

        match = feature_data[
            feature_data["feature_record_id"].eq(feature_id)
            & feature_data["service_option_id"].eq(
                base_result["service_option_id"]
            )
            & feature_data["feature_type"].eq("optional_add_on")
        ]

        if len(match) != 1:
            raise ValueError(
                "Add-on does not uniquely match the selected option: "
                + feature_id
            )

        feature = match.iloc[0]
        unit = feature["charging_unit"]

        if feature["currency"] != base_result["currency"]:
            raise ValueError("Currency mismatch.")

        line_cost = None
        status = "Unknown: charging unit not supported or confirmed"

        if unit == "per_bottle":
            price = float(feature["additional_unit_price"])

            if not math.isfinite(price) or price <= 0:
                raise ValueError("Invalid add-on price.")

            line_cost = round(price * quantity, 2)
            known_addon_total += line_cost
            status = "Calculated from supplied text; confirmation required"
        else:
            unresolved.append(feature_id)

        rows.append({
            "feature_record_id": feature_id,
            "feature_name": feature["feature_name"],
            "quantity": quantity,
            "charging_unit": unit,
            "provisional_line_cost": line_cost,
            "status": status
        })

    known_subtotal = round(
        float(base_result["provisional_base_cost"])
        + known_addon_total,
        2
    )

    budget = float(base_result["vendor_budget"])

    if known_subtotal > budget:
        budget_status = "Failed: known subtotal exceeds budget"
    elif unresolved:
        budget_status = "Unknown: selected add-on costs remain unresolved"
    else:
        budget_status = (
            "Known selected costs within budget; other charges unverified"
        )

    summary = {
        "service_option_id": base_result["service_option_id"],
        "currency": base_result["currency"],
        "provisional_base_cost":
            base_result["provisional_base_cost"],
        "known_selected_addon_cost": round(known_addon_total, 2),
        "known_cost_subtotal": known_subtotal,
        "balance_before_unresolved_charges": round(
            budget - known_subtotal, 2
        ),
        "unresolved_addon_ids": unresolved,
        "budget_check": budget_status,
        "price_period_check": base_result["price_period_check"],
        "availability_check": base_result["availability_check"],
        "venue_check": base_result["venue_check"],
        "overall_status": "Provisional calculation, not booking approval"
    }

    return {
        "addon_details": pd.DataFrame(rows),
        "summary": summary
    }
```

```python
import math
from itertools import product

import numpy as np
import pandas as pd


# =========================================================
# 1. Individual vendor candidate evaluation
# =========================================================
def evaluate_vendor_candidates(requirements, data):
    if not isinstance(requirements, dict):
        raise ValueError("requirements must be a dictionary.")

    required_columns = [
        "business_id",
        "name",
        "city",
        "state",
        "service_categories",
        "risk_level",
        "mean_review_rating",
        "collected_review_count",
        "rating_std",
    ]

    missing_columns = [
        column for column in required_columns
        if column not in data.columns
    ]

    if missing_columns:
        raise ValueError(
            "Missing columns: " + ", ".join(missing_columns)
        )

    city = str(requirements.get("city") or "").strip()
    state = str(requirements.get("state") or "").strip()
    service_inputs = requirements.get("services")

    if not city or not state:
        raise ValueError("Enter event city and state.")

    if not isinstance(service_inputs, dict) or not service_inputs:
        raise ValueError(
            "services must be a non-empty dictionary."
        )

    services = [str(service).strip() for service in service_inputs]

    if any(not service for service in services):
        raise ValueError("Service names cannot be blank.")

    if len(set(services)) != len(services):
        raise ValueError("Duplicate service names after trimming.")

    exclude_high = requirements.get(
        "exclude_high_review_labels", True
    )

    if not isinstance(exclude_high, bool):
        raise ValueError(
            "exclude_high_review_labels must be True or False."
        )

    # Proposed default weights, not validated research findings.
    default_weights = {
        "rating": 0.60,
        "review_volume": 0.20,
        "review_consistency": 0.20,
    }

    priorities = requirements.get("priorities") or {}

    if not isinstance(priorities, dict):
        raise ValueError("priorities must be a dictionary.")

    values = [
        priorities.get(key) for key in default_weights
    ]

    if all(value is None for value in values):
        weights = default_weights.copy()
        weight_source = "Proposed default weights"

    elif any(value is None for value in values):
        raise ValueError(
            "Supply all three priority weights or leave all unset."
        )

    else:
        weights = {
            key: float(priorities[key])
            for key in default_weights
        }

        if any(
            not math.isfinite(value) or value < 0
            for value in weights.values()
        ):
            raise ValueError(
                "Weights must be finite and non-negative."
            )

        total = sum(weights.values())

        if total <= 0:
            raise ValueError("Priority weights cannot all be zero.")

        weights = {
            key: value / total
            for key, value in weights.items()
        }

        weight_source = "User priority weights"

    work = data.copy()

    ids = work["business_id"].astype("string").str.strip()

    invalid_ids = (
        ids.isna()
        | ids.fillna("").eq("")
        | ids.fillna("").str.startswith("#")
    )

    if invalid_ids.any():
        raise ValueError("Missing or corrupted business IDs.")

    if ids.duplicated().any():
        raise ValueError("Duplicate business IDs.")

    work["business_id"] = ids

    city_match = (
        work["city"].fillna("").astype(str)
        .str.strip().str.casefold()
        .eq(city.casefold())
    )

    state_match = (
        work["state"].fillna("").astype(str)
        .str.strip().str.casefold()
        .eq(state.casefold())
    )

    local = work.loc[city_match & state_match].copy()

    allowed_labels = (
        ["Low", "Medium"]
        if exclude_high
        else ["Low", "Medium", "High"]
    )

    local = local.loc[
        local["risk_level"].isin(allowed_labels)
    ].copy()

    numeric_columns = [
        "mean_review_rating",
        "collected_review_count",
        "rating_std",
    ]

    for column in numeric_columns:
        local[column] = pd.to_numeric(
            local[column], errors="coerce"
        )

    finite_values = pd.Series(
        np.isfinite(
            local[numeric_columns].to_numpy(dtype=float)
        ).all(axis=1),
        index=local.index,
    )

    valid_features = (
        local[numeric_columns].notna().all(axis=1)
        & finite_values
        & local["mean_review_rating"].between(1, 5)
        & local["collected_review_count"].ge(1)
        & local["rating_std"].ge(0)
    )

    invalid_feature_count = int((~valid_features).sum())
    local = local.loc[valid_features].copy()

    result_columns = [
        "business_id",
        "name",
        "assigned_service",
        "city",
        "state",
        "risk_level",
        "mean_review_rating",
        "collected_review_count",
        "rating_std",
        "event_ranking_score",
        "eligibility_status",
        "selection_reason",
        "checks_not_evaluated",
    ]

    evaluated_parts = []
    missing_services = []

    for service in services:
        service_match = (
            local["service_categories"].fillna("").astype(str)
            .apply(
                lambda value: service in {
                    item.strip()
                    for item in value.split(";")
                }
            )
        )

        candidates = local.loc[service_match].copy()

        if candidates.empty:
            missing_services.append(service)
            continue

        rating_component = (
            candidates["mean_review_rating"] / 5.0
        )

        # Relative to this local service candidate pool.
        maximum_reviews = (
            candidates["collected_review_count"].max()
        )

        volume_component = (
            np.log1p(candidates["collected_review_count"])
            / np.log1p(maximum_reviews)
        )

        consistency_component = (
            1.0 / (1.0 + candidates["rating_std"])
        )

        candidates["event_ranking_score"] = 100.0 * (
            weights["rating"] * rating_component
            + weights["review_volume"] * volume_component
            + weights["review_consistency"]
            * consistency_component
        )

        candidates["assigned_service"] = service
        candidates["eligibility_status"] = "Candidate only"

        candidates["selection_reason"] = (
            "Service category and business city/state match; "
            "review label meets the selected preference."
        )

        candidates["checks_not_evaluated"] = (
            "Event-type suitability; budget; guest capacity; "
            "date/time availability; service-specific requirements; "
            "service area; venue; operational compatibility"
        )

        evaluated_parts.append(
            candidates[result_columns]
        )

    if evaluated_parts:
        evaluated = pd.concat(
            evaluated_parts, ignore_index=True
        ).sort_values(
            ["assigned_service", "event_ranking_score", "business_id"],
            ascending=[True, False, True],
        ).reset_index(drop=True)
    else:
        evaluated = pd.DataFrame(columns=result_columns)

    return {
        "candidates": evaluated,
        "missing_services": missing_services,
        "weights": weights,
        "weight_source": weight_source,
        "invalid_feature_rows_excluded": invalid_feature_count,
    }


# =========================================================
# 2. Dynamic candidate package generation
# =========================================================
def generate_dynamic_packages(
    requirements,
    vendor_data,
    top_per_service=3,
    max_combinations=10000,
):
    for parameter_name, value in [
        ("top_per_service", top_per_service),
        ("max_combinations", max_combinations),
    ]:
        if (
            not isinstance(value, int)
            or isinstance(value, bool)
            or value < 1
        ):
            raise ValueError(
                parameter_name + " must be a positive integer."
            )

    if not isinstance(requirements, dict):
        raise ValueError("requirements must be a dictionary.")

    number_of_packages = requirements.get(
        "number_of_packages", 3
    )

    if (
        not isinstance(number_of_packages, int)
        or isinstance(number_of_packages, bool)
        or number_of_packages < 1
    ):
        raise ValueError(
            "number_of_packages must be a positive integer."
        )

    evaluation = evaluate_vendor_candidates(
        requirements,
        vendor_data,
    )

    common_result = {
        "missing_services": evaluation["missing_services"],
        "weights": evaluation["weights"],
        "weight_source": evaluation["weight_source"],
        "invalid_feature_rows_excluded":
            evaluation["invalid_feature_rows_excluded"],
    }

    if evaluation["missing_services"]:
        return {
            **common_result,
            "status": "Incomplete: missing eligible candidates",
            "packages": [],
            "combinations_considered": 0,
            "shared_vendor_combinations_excluded": 0,
        }

    services = [
        str(service).strip()
        for service in requirements["services"]
    ]

    candidates = evaluation["candidates"]
    pools = []

    for service in services:
        pool = candidates.loc[
            candidates["assigned_service"].eq(service)
        ].sort_values(
            ["event_ranking_score", "business_id"],
            ascending=[False, True],
        )

        pool = pool.drop_duplicates("business_id")
        pool = pool.head(top_per_service)

        pools.append(pool.to_dict("records"))

    combination_count = math.prod(
        len(pool) for pool in pools
    )

    if combination_count > max_combinations:
        raise ValueError(
            f"{combination_count} combinations exceed the "
            f"{max_combinations} limit. Reduce top_per_service."
        )

    packages = []
    seen_signatures = set()
    excluded_shared_vendor_combinations = 0

    for combination in product(*pools):
        business_ids = [
            item["business_id"] for item in combination
        ]

        # Prototype policy: distinct vendor per service.
        # This is not proof that multi-service delivery is impossible.
        if len(set(business_ids)) != len(business_ids):
            excluded_shared_vendor_combinations += 1
            continue

        signature = tuple(
            (item["assigned_service"], item["business_id"])
            for item in combination
        )

        if signature in seen_signatures:
            continue

        seen_signatures.add(signature)

        selected = [
            {
                "service": item["assigned_service"],
                "business_id": str(item["business_id"]),
                "vendor_name": item["name"],
                "review_based_label": item["risk_level"],
                "event_ranking_score": float(
                    item["event_ranking_score"]
                ),
            }
            for item in combination
        ]

        # Equal service weights for this prototype.
        package_score = sum(
            item["event_ranking_score"]
            for item in selected
        ) / len(selected)

        packages.append({
            "selected_vendors": selected,
            "package_score": package_score,
            "status": "Candidate package only",
            "checks": {
                "required_service_categories": "Matched",
                "business_city_state": "Matched",
                "review_label_preference": "Matched",
                "event_type_suitability": "Not evaluated",
                "budget": "Not evaluated",
                "guest_capacity": "Not evaluated",
                "date_time_availability": "Not evaluated",
                "service_specific_requirements": "Not evaluated",
                "service_area": "Not evaluated",
                "venue": "Not evaluated",
                "operational_compatibility": "Not evaluated",
            },
            "explanation": (
                "Selected from matching service categories and "
                "business locations, using the selected review-label "
                "preference and ranking priorities. "
                "Other feasibility requirements remain unverified."
            ),
            "_signature": signature,
        })

    packages.sort(
        key=lambda package: (
            -package["package_score"],
            package["_signature"],
        )
    )

    best_packages = packages[:number_of_packages]

    for rank, package in enumerate(best_packages, start=1):
        package["package_rank"] = rank
        package["package_score"] = round(
            float(package["package_score"]), 2
        )
        del package["_signature"]

    return {
        **common_result,
        "status": (
            "Candidate packages generated"
            if best_packages
            else "No package under distinct-vendor policy"
        ),
        "packages": best_packages,
        "combinations_considered": combination_count,
        "shared_vendor_combinations_excluded":
            excluded_shared_vendor_combinations,
    }


print(
    "Evaluator ready:",
    callable(evaluate_vendor_candidates)
)
print(
    "Generator ready:",
    callable(generate_dynamic_packages)
)
```
